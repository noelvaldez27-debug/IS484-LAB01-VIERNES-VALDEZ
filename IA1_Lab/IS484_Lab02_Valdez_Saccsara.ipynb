## **Problema del repartidor**

#### **Parte A: Completar la cha del problema para el siguiente caso:**
**Estado inicial** = El repartidor se encuentra en el Almacén
**Estado objetivo** = El repartidor llega a la Zona Norte
**Acciones posibles** = Moverse de un punto a otro
**Costo de cada acción** = El costo de accion va ser el mismo

#### **Parte B: Representar el mapa de la ciudad como un grafo en Python**

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

#### **Implementación de BFS búsqueda en amplitud**

```python
# Se define la funcion
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """

# la cola de camino inicia con en el estado inicial, "Almacen"
    cola = [[inicio]]                    # cola de caminos, empieza con el estado inicial
    visitados = set()                    # En la variable visitados, se crea un conjunto de datos que va almacenar los elementos 

# Repite mientras queden caminos por revisar en la cola
    while cola:
        camino = cola.pop(0)             # Saca el "primer" camino de la cola asi se explora por niveles
        nodo_actual = camino[-1]         # Toma el ultimo nodo del camino, donde esta el repartidor ahora

# Si ese nodo es el objetivo, devuelve el camino
        if nodo_actual == objetivo:      
            return camino

        if nodo_actual not in visitados:             # Solo expande el nodo si aun no fue explorado
            visitados.add(nodo_actual)               # Marca el nodo como visitado
            for vecino in grafo[nodo_actual]:        # Recorre los vecinos conectados directamente al nodo actual
                nuevo_camino = camino + [vecino]     # Crea un camino nuevo, el actual mas el vecino al final
                cola.append(nuevo_camino)            # Agrega el nuevo camino al "final" de la cola


    return None      # Si la cola se vacio sin llegar a la meta, no existe camino


# Ejecuta la busqueda de Almacen a ZonaNorte
resultado = bfs(mapa, "Almacen", "ZonaNorte")

# Muestra el camino encontrado 
print("Camino encontrado:", resultado)
```

#### **Vectores y matrices con NumPy**

#### Parte A: Representar cuatro observaciones de un dataset simple

```python
# Importamos NumPy y le nombramos con "np"
import numpy as np

# Cada observacion es un vector: [temperatura, humedad]
obs1 = np.array([22.5, 60])  # temperatura, humedad
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

# Juntamos los 4 vectores en una matriz: cada fila es una observacion
dataset = np.array([obs1, obs2, obs3, obs4])

# Mostramos la matriz completa
print(dataset)

# # .shape devuelve filas y columnas
print("Forma de la matriz (filas, columnas):", dataset.shape)
```

#### Parte B: Acceder a una la y a una columna especíca de la matriz

```python
# Muestra la fila
print("Primera observacion (fila 0):", dataset[0])

# Muestra la columna
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```
